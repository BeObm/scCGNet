```python
import sys, glob, os
import numpy as np, pandas as pd
import matplotlib; matplotlib.use("Agg")
import matplotlib.pyplot as plt
import umap

# usage: python umap_plot.py embedding_80.npy types_80.txt [more pairs...]
#    or: python umap_plot.py            (auto-pairs embedding_*.npy with types_*.txt)

def run(emb_path, types_path):
    emb = np.load(emb_path)
    df = pd.read_csv(types_path, index_col=0)
    u = umap.UMAP(random_state=42).fit_transform(emb)
    fig, ax = plt.subplots(1, 2, figsize=(14, 6))
    for a, col in zip(ax, ["True", "Pred"]):
        sc = a.scatter(u[:, 0], u[:, 1], c=df[col], cmap="tab20", s=6)
        a.set_title(col); a.set_xlabel("UMAP-1"); a.set_ylabel("UMAP-2")
        fig.colorbar(sc, ax=a)
    out = os.path.splitext(os.path.basename(emb_path))[0] + "_umap.png"
    fig.tight_layout(); fig.savefig(out, dpi=200); plt.close(fig)
    print("wrote", out)

if len(sys.argv) > 1:
    args = sys.argv[1:]
    for i in range(0, len(args), 2):
        run(args[i], args[i + 1])
else:
    for emb in sorted(glob.glob("embedding_*.npy")):
        run(emb, emb.replace("embedding_", "types_").replace(".npy", ".txt"))
```
